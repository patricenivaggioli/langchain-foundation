# 1.4 Memory

By default, Large Language Models are stateless, meaning they treat each interaction as an independent event. This limitation prevents them from maintaining context in a conversation.

In this notebook, we will explore:
1.  **Stateless Interactions**: Demonstrating how models fail to remember previous inputs.
2.  **Adding Memory**: Implementing mechanisms to store and retrieve conversation history.
3.  **Stateful Conversations**: Building an agent that can recall user details and context across multiple turns.

```python
from dotenv import load_dotenv

load_dotenv()
```

## 1.4.1 No memory

```python
import os
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model=os.getenv("LLM_MODEL"),
    api_key=os.getenv("LLM_API_KEY"),
    base_url=os.getenv("LLM_BASE_URL")
)


agent = create_agent(
    model=model
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Seán and my favourite colour is green")

response = agent.invoke(
    {"messages": [question]} 
)
```

```python
from pprint import pprint

pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]} 
)

pprint(response)
```

## 1.4.2 Memory

```python
from langgraph.checkpoint.memory import InMemorySaver  


agent = create_agent(
    model=model,
    checkpointer=InMemorySaver(),  
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Seán and my favourite colour is green")
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [question]},
    config,  
)
```

```python
pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]},
    config,  
)

pprint(response)
```
